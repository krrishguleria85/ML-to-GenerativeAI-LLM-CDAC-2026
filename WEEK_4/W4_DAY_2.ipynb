```python
# Transformer: input text sequence -> tokenisation -> embedding layer then goes to embedding stack (n layers) -> [ decoder stack ] <- -> linear layer -> activation i.e, softmax -> output generation.

# the arch. of encoder and decoder:

# inside the encoder stack: <--[ 1. add and normalise 2. feed forward network --back to (1)-- 3.add and normalise 4. multi head self attention --back to (3) ] <--- where, multi head self attention: maps rel between all the tokens simultaneously, feed forward network: processes individual tokens representation independently,
# layer normalisation (4): stabilises parallel training , residual connections (3): prevents vanishing gradient problem

# inside the decoder stack

# -------------------------

# complete attention all in need research paper formula is Attention (Q, K ,V) = softmax(Q K ^T / ROOT OF dk) V where: Q is query, k is key, v is the value..

# step 1: input projections: there are the input embedding matrix X, which shape N x d and then it has three types: W^Q, W^k, W^t..
# step 2: dot product attention: act as a similarity of the tokens. A higher score indicates higher relevance between the tokens. Score = Q . K^T
#step 3: scaled dot product attention: QK^T / root d ---> this ensures the query to be scaled and smooth in progression.
# step 4: softmax normalisation : it converts arbitary raw sources into probability distribution
#step 5: Context vector generation: it picks most important tokens to get further prog..
```

```python
# CODE

#import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import nltk
import re #regular expression
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import wordnet
from nltk import pos_tag

print("Libraries imported successfully!!!")
```

```python
# Section 1: Download nltk resources

resources = [
    'punkt',
    'punkt_tab',
    'averaged_perceptron_tagger',
    'averaged_perceptron_tagger_eng',
    'wordnet',
    'omw-1.4'
]
for resource in resources:
  nltk.download(resource)
```

```python
# Section 2: Basic Text Preprocessing

text = 'Natural Language Processing, enables computers to understand the user language in a efficient manner.'
print("Original text: ", text)

tokens = word_tokenize(text)
print("\n Tokens: \n ",tokens)
```

```python
#Section 3: rule based system: sentiment analysis

sentence = "I love this movie"

if "love" in sentence.lower():
  sentiment ="Positive"
elif "hate" in sentence.lower():
  sentiment ="Negative"
else:
  sentiment = "neutral"

print("Sentence:", sentence)
print("Sentiment: ", sentiment)
```

```python
# grammer based parsing using pos tagging

sentence = "The cat sat on the mat"

tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)

print("POS Tags:")
print(pos_tags)
```

```python
# sentence structural analysis

sentence = "Natural Language Processing is good"
tokens = word_tokenize(sentence)
tagged_words = pos_tag(tokens)

for word, tag in tagged_words:
  print(f"{word} --> {tag}")
```

```python
# lexicons and dictonaries using wordnet

word ="computer"
sysnets = wordnet.synsets(word)

print("Defination: ")
for syn in sysnets[:3]:
  print(syn.definition())
```

```python
# find synonyms and antonymes

word = "good"

synonyms = set()
antonyms = set()

for syn in wordnet.synsets(word):
  for lemma in syn.lemmas():
    synonyms.add(lemma.name())

    if lemma.antonyms():
      antonyms.add(lemma.antonyms()[0].name())

print("Synonyms:")
print(list(synonyms)[:10])
```

```python
#import libraries

import numpy as np
import pandas as pd

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

import nltk
from nltk.util import ngrams
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
#  sample corpus

documents = [
    "I love Natural Language processing",
    "NLP is fascinating",
    "Machine learning powers NLP",
    "Deep Learning improves NLP Models"
]

for i , doc in enumerate(documents, start=1):
  print(f" Document {i} : {doc}")
```

```python
# BAG OF WORDS [ document to vector conversion ]

vectorizer = CountVectorizer()

X = vectorizer.fit_transform(documents)
bow_df = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out()
)
print(bow_df)
```

```python
# vocalbulary analysis
vocabulary = vectorizer.get_feature_names_out()

print("Vocabulary:\n")
for word in vocabulary:
  print(word)
```

```python
# word freq analysis

word_counts = np.sum(X.toarray(), axis=0)

freq_df = pd.DataFrame(
    {
        "Word": vocabulary,
        "Frequency": word_counts
    }
)
freq_df = freq_df.sort_values(
    by = "Frequency",
    ascending = False
)
print(freq_df)
```

```python
#TF - IDF representation

tfidf_vectorizer = TfidfVectorizer()

X_tfidf = tfidf_vectorizer.fit_transform(documents)

tfidf_df =pd.DataFrame(
    X_tfidf.toarray(),
    columns= tfidf_vectorizer.get_feature_names_out()
)
print(tfidf_df.round(3))
```
