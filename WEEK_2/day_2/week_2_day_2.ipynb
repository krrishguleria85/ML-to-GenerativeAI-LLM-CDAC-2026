```python
#Linear Regression -> Supervised Learning

#It means:  linear refers to line and the regression means it tells the relationship between the data points.
# y = m X + c
# where, y is the dependent variable (target), X is the independent variable (input features)

#error calc = actual data point - predicted datapoint (E = y - y hat )
#smaller error meand better prediction

#when we reducing the error it called optimisation, and to reduce error we use various key metrices that include: root mean square, mean square error, mean absolute error, R^2 -> coefficient of determinant..

```

```python
#The cost functions: Mean Squared error:
# MSE = 1/n sigma (y actual - y predicted)^2 where; (y actual - y predicted) -> difference between actual and predicted values , and sigma uses to adds all individual errors together.


# R^2 (coeff. of determinant) = 1 - sigma (y actual - y predicted)^2 / sigma (y actual - y predicted)^2 || 1 - sum of square of residual / sum of square of total
#In R2, it contains two line include mean line and regression line that take the error distance between the data point.
```

```python
#expanding multiple linear regression

# same as single linear regression but it contains multiple coefficient and multiple input features but output of predicted is one.
# y (predicted) = b not (intercept) + [ b1 X1 + b2 X2 + ... + bn Xn ] where, b1 , b2 -> coefficient ; X1 .. Xn -> input features
```

LASSO REGRESSION -> least absolute shrinkage and selection operator -> used to reduces the overfitting

```python
# It is used to overcome the underfitting and overfitting techniques

# real world problems in linear regression: 1. as in real world datasets there are too much noise and 2. model tries to hard to fit every data point and 3. also results in extreme mathematical instability..

# understanding model complexity -> to calculate the model complexity these steps are : too many features, explding coefficient and overfitting

# why large coefficient are dangerous ? --> example: y predicted = 5 + 500  x1 [ if x1 = 2.0 then prediction = 1005 and if x1 is 2.1 then prediction is 1055]
```

```python
# Regularization : It keeps the model simple and stable by intentionally restricting the mathematical size of the best restricting the mathematical size of the coefficients.

#types of regularization:
#L1 (Lasso regression): uses the coefficients value of |b|
#L2 (ridge regression): uses the square coefficients -> b^2

#Both types do same penalty, but they do it in fundamentally diff. mathematical ways..
```

```python
# core idea of lasso regression:

# minimizes the model complexity that means to reduce the coefficients value that present to be high (need to shrink)
# minimize the error prediction
#centre part is optimal lasso value

#---------------------------------------
# lasso regression cost function: 1/n sigma (y actual - y predicted)^2  +  ( lambda sigma |bj| ) -> this is the penalty function
# ridge regression cost function: 1/n sigma (y actual - y predicted)^2  +  ( lambda sigma bj^2 ) -> this is the penalty function
#---------------------------------------
#**also search this **
```

```python
# elastic net cost function : combines both L1 and L2 penalty

# Cost = 1/n sigma (y actual - y predicted)^2 + ( lambda sigma |bj| ) + ( lambda sigma bj^2 )
```

```python
# Tuning complexity : lambda parameters

# A: small penalty : high risk of Overfitting
# B: strong penalty : high risk of underfitting
```
