```python
#Q3. Create two random matrices of size 4x4.
#Perform:
#- Addition
#- Subtraction
#- Element-wise multiplication
#- Matrix multiplication


mat1 = np.random.randint(1,51, size=(4,4))
mat2 = np.random.randint(1,51, size=(4,4))
print("Matrix 1:\n", mat1)
print("\n")
print("Matrix 2:\n", mat2)
print("\n")

print("Addition of two matrix: \n")
add_mat = np.add(mat1, mat2)
print(add_mat)
print("\n")

print("Subtraction of two matrix: \n")
sub_mat = np.subtract(mat1, mat2)
print(sub_mat)
print("\n")

print("Element-wise multiplication of two matrix: \n")
multiply_element_mat = np.multiply(mat1, mat2)
print(multiply_element_mat)
print("\n")

print("Matrix multiplication of two matrix: \n")
matrix_multiplication = np.dot(mat1, mat2)
print(matrix_multiplication)
print("\n")
```

```python
#Q26. Load a dataset and create:
#- Heatmap
#- Pairplot
#Explain observations from both plots.


iris = sns.load_dataset("iris")

plt.figure(figsize=(10,8))
sns.heatmap(iris.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Heatmap")
plt.show()
```

```python
sns.pairplot(iris, hue="species")
plt.title("Pairplot")
plt.show()

#Observations:
#1. Heatmap
#Petal Length and Petal Width have a very strong positive correlation.
#Sepal Length is moderately correlated with petal measurements and the sepal Width has a weak correlation (in negative) with the other features.

#2. Pairplot
#The species setosa is clearly separated from the other two species.
#Versicolor and virginica overlap slightly but can still be distinguished using petal measurements.
#Petal length and petal width are the best features for classifying the flower species.
```

```python
#Q30. Create two matrices and perform:
#- Determinant
#- Inverse
#- Eigenvalues
#- Eigenvectors
#using scipy.linalg.
#Visualize and interpret the results.


mat1 = np.array([[4,2],
                 [1,3]])

mat2 = np.array([[2,1],
                 [5,4]])

print("Matrix 1: \n",mat1)
print("Matrix 2: \n",mat2)
print()

print("Determinant of Matrix 1: \n", linalg.det(mat1))
print("Determinant of Matrix 2: \n", linalg.det(mat2))
print()



print("Inverse of Matrix 1:\n", linalg.inv(mat1))
print("Inverse of Matrix 2:\n", linalg.inv(mat2))
print()

#- Eigenvalues, Eigenvectors
eigen_values, eigen_vectors = linalg.eig(mat1)

print("Eigenvalues:")
print(eigen_values)
print()

print("Eigenvectors:")
print(eigen_vectors)
print()

# Visualization
plt.bar(["Eigenvalue 1", "Eigenvalue 2"],
    eigen_values.real,
    color="skyblue",
    edgecolor="black"
)

plt.title("Eigenvalues of Matrix 1")
plt.ylabel("Value")
plt.show()
```
