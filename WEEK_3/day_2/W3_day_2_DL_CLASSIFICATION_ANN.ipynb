```python
fig, ax = plt.subplots(
    figsize=(10,5)
)
ax.axis("off")

# Cell Body
cell = plt.Circle(
    (0.5,0.5),
    0.08,
    fill=True
)
ax.add_patch(cell)

# Dendrites
for y in [0.2,0.35,0.5,0.65,0.8]:

    ax.plot(
        [0.1,0.42],
        [y,0.5],
        linewidth=2
    )

# Axon
ax.plot(
    [0.58,0.9],
    [0.5,0.5],
    linewidth=3
)

ax.text(
    0.02,
    0.85,
    "Dendrites",
    fontsize=12
)

ax.text(
    0.45,
    0.62,
    "Cell Body",
    fontsize=12
)

ax.text(
    0.82,
    0.6,
    "Axon",
    fontsize=12
)

plt.title(
    "Biological Neuron"
)

plt.show()
```

```python
fig, ax = plt.subplots(
    figsize=(10,5)
)

ax.axis("off")

# Inputs

inputs_y = [0.2,0.5,0.8]

for y in inputs_y:

    ax.scatter(
        0.2,
        y,
        s=300
    )

# Neuron

ax.scatter(
    0.6,
    0.5,
    s=2000
)

# Output

ax.scatter(
    0.9,
    0.5,
    s=300
)

# Connections

for y in inputs_y:

    ax.plot(
        [0.2,0.6],
        [y,0.5]
    )

ax.plot(
    [0.6,0.9],
    [0.5,0.5]
)

ax.text(0.15,0.2,"X1")
ax.text(0.15,0.5,"X2")
ax.text(0.15,0.8,"X3")

ax.text(
    0.55,
    0.65,
    "Neuron"
)

ax.text(
    0.92,
    0.5,
    "Y"
)

plt.title(
    "Artificial Neuron"
)

plt.show()
```

```python
# tanh , range : (-1,1)
#use case: hidden layers

import numpy as np
import plotly.express as px

x = np.linspace(-10,10,1000)
y = np.tanh(x)

fig = px.line(x=x, y=y, title="Tanh Activation Function")
fig.show()
```

```python
#relu -> range: [0,+infinity)

x = np.linspace(-10,10,1000)
y=np.maximum(0,x)

fig = px.line(x=x, y=y, title="RelU activation function")
fig.show()
```

```python
# LEAKY RELU: range: (- to + infinity)

x = np.linspace(-10,10,1000)
y = np.where(
    x>0, x, 0.01*x
)
fig = px.line(
    x=x, y=y, title="Leaky RelU"
)
fig.show()
```

```python
# SOFTMAX : range: (0,1)
# Sum of prob = 1 and use case is multi class classification
scores = np.array([
    2.5,1.2,0.8,3.4
])
softmax = np.exp(scores)/np.sum(
    np.exp(scores)
)
print("SoftMax Probabilities: ")
print(softmax)

import pandas as pd
df=pd.DataFrame({
    "Class":[
        "Class A",
        "Class B",
        "Class C",
        "Class D"
    ],
    "Probability":softmax
})
fig = px.bar(df, x="Class", y="Probability", title="Softmax Prob. Distribution")
fig.show()
```

```python
# GELU (use case: transformer, LLM)
#range: (- to + infinity)

x = np.linspace(-10,10,1000)

gelu =(
    0.5*x*
    (
        1+
        np.tanh(
            np.sqrt(2/np.pi)*
            (
                x+
                0.044715*x**3
            )
        )
    )
)
fig = px.line(
    x=x, y=gelu, title="GELU Activation Function"
)
fig.show()

print("Range: (- to + infinity)")
print("Used in GPT, LLM")
```

```python
# BINARY STEP ACTIVATION FUNCTION
#range: {0,1}
#use case= perceptron

x = np.linspace(-10,10,1000)
y=np.where(
    x>=0,
    1,
    0
)
fig = px.line(x=x, y=y, title="Binary Step Act. function")
fig.show()
```

```python
# linear activation

x = np.linspace(-10,10,1000)
y=x

fig = px.line(x=x, y=y, title="linear Act. function")
fig.show()
```

```python
# parametric RelU and range: (- to + infinity)
#use case is deep learning networks
#alpha is differentiable and changing...

alpha=0.25
x=np.linspace(-10,10,1000)
y = np.where(
    x>0,x, alpha*x
)
fig = px.line(x=x, y=y, title="Parametric RelU")
fig.show()
```

```python
alpha=0.01
x=np.linspace(-10,10,1000)
y = np.where(
    x>0,x, alpha*x
)
fig = px.line(x=x, y=y, title="Parametric RelU")
fig.show()
```

```python
alpha=1
x=np.linspace(-10,10,1000)
y = np.where(
    x>0,x, alpha*x
)
fig = px.line(x=x, y=y, title="Parametric RelU")
fig.show()
```

```python
alpha=1.5
x=np.linspace(-10,10,1000)
y = np.where(
    x>0,x, alpha*x
)
fig = px.line(x=x, y=y, title="Parametric RelU")
fig.show()
```

```python
# ELU -> Exponential linear unit
#range: (-1 , + infinity)
# solves dead relu problem

alpha = 1
x=np.linspace(-10,10,1000)
y=np.where(
    x>0,x, alpha*(np.exp(x)-1)
)
fig = px.line(x=x, y=y, title="ELU")
fig.show()
```

```python
#ELU but alpha is -1

alpha = -1
x=np.linspace(-10,10,1000)
y=np.where(
    x>0,x, alpha*(np.exp(x)-1)
)
fig = px.line(x=x, y=y, title="ELU")
fig.show()
```

```python
# SELU -> scaled exponential linear unit
#range: (- to + infinity)

alpha = -2.15556
scale = -1.06606

x=np.linspace(-10,10,1000)
y=scale*np.where(
    x>0,x, alpha*(np.exp(x)-1)
)
fig = px.line(x=x, y=y , title="SELU")
fig.show()
```

```python
# SOFTPLUS
#range: (0, +infinity)
#use case: Smooth RelU Alternative

x=np.linspace(-10,10,1000)
y=np.log(
    1+np.exp(x)
)
fig = px.line(x=x,y=y, title="Softplus Act.. Function")
fig.show()
```

```python
#Softsign
#range: (-1,1)
#use case: alternative to tanh

x=np.linspace(-10,10,1000)
y=x/(
    1+np.abs(x)
)
fig = px.line(x=x,y=y, title="Soft Sign Act.. Function")
fig.show()
```

```python
#load dataset

from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()

X = pd.DataFrame(cancer.data, columns = cancer.feature_names)
y= pd.Series(cancer.target, name="Target")

print(X.shape)
print(y.shape)
print("\nHead of X:\n",X.head())
print("\nHead of Y: \n",y.head())
```

```python
# feature scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print(X_scaled[:5])

```

```python
# train test split
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
# ANN MODEL


from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

#input layer
model.add(
    Dense(
        units=30,
        activation = "relu",
        input_shape=(30,)
    )
)

#hidden layer 1:
model.add(
    Dense(
        16,
        activation= "relu")
)

#hidden layer 2:
model.add(
    Dense(8, activation= "relu")
)

#Output Layer:
model.add(
    Dense(1, activation= "sigmoid")
)

model.summary()
```

```python
# compile of model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
#train the model

history = model.fit(
    X_train,
    y_train,
    epochs = 50,
    batch_size =32,
    validation_split =0.2,
    verbose=1
)
```

```python
#draw the accuracy_curve
import plotly.graph_objects as go

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        y=history.history["accuracy"],
        name='Training Accuracy'
    )
)

fig.add_trace(
    go.Scatter(
        y=history.history["val_accuracy"],
        name='Validation Accuracy'
    )
)

fig.update_layout(
    title="Training vs Validation"
)
fig.show()
```

```python
#evaluation

loss, accuracy = model.evaluate(
    X_test,
    y_test
)
print("Accuracy:",accuracy)
```
