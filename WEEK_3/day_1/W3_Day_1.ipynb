```python
# ============================================================
# FEEDFORWARD NEURAL NETWORK
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    figsize=(12,6)
)

ax.axis("off")

inputs = np.linspace(
    1,
    4,
    4
)

hidden = np.linspace(
    1.5,
    3.5,
    3
)

output = [2.5]

# Input Layer

for y in inputs:

    ax.scatter(
        1,
        y,
        s=700
    )

# Hidden Layer

for y in hidden:

    ax.scatter(
        4,
        y,
        s=700
    )

# Output Layer

ax.scatter(
    7,
    output[0],
    s=900
)

# Connections

for i in inputs:

    for h in hidden:

        ax.plot(
            [1,4],
            [i,h]
        )

for h in hidden:

    ax.plot(
        [4,7],
        [h,output[0]]
    )

ax.text(1,0.3,"Input")
ax.text(4,0.3,"Hidden")
ax.text(7,0.3,"Output")

plt.title(
    "Feedforward Neural Network"
)

plt.show()
```

```python
# ============================================================
# FEEDFORWARD CALCULATION
# ============================================================

import numpy as np

X = np.array([
    2,
    3
])

W1 = np.array([
    [0.5,0.2],
    [0.3,0.8]
])

B1 = np.array([
    0.1,
    0.2
])

hidden = np.dot(
    X,
    W1
) + B1

hidden = np.maximum(
    0,
    hidden
)

print("Hidden Layer")

print(hidden)

W2 = np.array([
    [0.4],
    [0.6]
])

B2 = np.array([
    0.5
])

output = np.dot(
    hidden,
    W2
) + B2

print("Output")

print(output)
```

```python
# ============================================================
# SHALLOW NEURAL NETWORK
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    figsize=(12,6)
)

ax.axis("off")

layers = [
    4,
    5,
    2
]

positions = []

for layer_idx, layer_size in enumerate(layers):

    y_positions = np.linspace(
        -(layer_size-1)/2,
        (layer_size-1)/2,
        layer_size
    )

    current_layer = []

    for y in y_positions:

        ax.scatter(
            layer_idx*2,
            y,
            s=600
        )

        current_layer.append(
            (layer_idx*2,y)
        )

    positions.append(
        current_layer
    )

for i in range(
    len(positions)-1
):

    for p1 in positions[i]:

        for p2 in positions[i+1]:

            ax.plot(
                [p1[0],p2[0]],
                [p1[1],p2[1]]
            )

plt.title(
    "Shallow Neural Network"
)

plt.show()
```

```python
# ============================================================
# DEEP NEURAL NETWORK
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    figsize=(15,8)
)

ax.axis("off")

layers = [
    4,
    6,
    5,
    4,
    3,
    2
]

positions = []

for layer_idx, layer_size in enumerate(layers):

    y_positions = np.linspace(
        -(layer_size-1)/2,
        (layer_size-1)/2,
        layer_size
    )

    current_layer = []

    for y in y_positions:

        ax.scatter(
            layer_idx*2,
            y,
            s=500
        )

        current_layer.append(
            (layer_idx*2,y)
        )

    positions.append(
        current_layer
    )

for i in range(
    len(positions)-1
):

    for p1 in positions[i]:

        for p2 in positions[i+1]:

            ax.plot(
                [p1[0],p2[0]],
                [p1[1],p2[1]],
                linewidth=0.5
            )

plt.title(
    "Deep Neural Network"
)

plt.show()
```

```python
# ============================================================
# PERCEPTRON
# ============================================================

import numpy as np
import pandas as pd
import plotly.express as px

# Input Features

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

# AND Gate Labels

y = np.array([
    0,
    0,
    0,
    1
])

# Initial Weights

weights = np.array([
    0.5,
    0.5
])

bias = -0.7

# ============================================================
# STEP ACTIVATION FUNCTION
# ============================================================

def step_function(x):

    if x >= 0:

        return 1

    return 0

# ============================================================
# PERCEPTRON OUTPUT
# ============================================================

for i in range(len(X)):

    weighted_sum = (
        np.dot(
            X[i],
            weights
        ) + bias
    )

    prediction = step_function(
        weighted_sum
    )

    print(
        X[i],
        prediction
    )

print("Perceptron Completed")
```

```python
# ============================================================
# SINGLE LAYER PERCEPTRON (SLP)
# ============================================================

import numpy as np
import pandas as pd

from sklearn.linear_model import Perceptron

import plotly.express as px

# ============================================================
# DATASET
# ============================================================

X = np.array([
    [0,0],
    [0,1],
    [1,0],
    [1,1]
])

y = np.array([
    0,
    0,
    0,
    1
])

# ============================================================
# MODEL TRAINING
# ============================================================

model = Perceptron()

model.fit(
    X,
    y
)

# ============================================================
# PREDICTION
# ============================================================

predictions = model.predict(X)

print(predictions)

# ============================================================
# WEIGHTS
# ============================================================

print("Weights")

print(model.coef_)

print("Bias")

print(model.intercept_)


print("SLP Completed")
```

```python
# ============================================================
# MLP CALCULATION (FORWARD PROPAGATION)
# ============================================================

import numpy as np

# ============================================================
# INPUT LAYER
# ============================================================

X = np.array([
    2,
    3
])

print("Inputs")

print(X)

# ============================================================
# INPUT -> HIDDEN WEIGHTS
# ============================================================

W1 = np.array([
    [0.5, 0.2],
    [0.3, 0.8]
])

print("\nInput-Hidden Weights")

print(W1)

# ============================================================
# HIDDEN BIAS
# ============================================================

B1 = np.array([
    0.1,
    0.2
])

print("\nHidden Bias")

print(B1)

# ============================================================
# HIDDEN LAYER CALCULATION
# ============================================================

hidden_input = (
    np.dot(X, W1)
    + B1
)

print("\nHidden Layer Input")

print(hidden_input)

# ============================================================
# RELU ACTIVATION
# ============================================================

hidden_output = np.maximum(
    0,
    hidden_input
)

print("\nHidden Layer Output")

print(hidden_output)

# ============================================================
# HIDDEN -> OUTPUT WEIGHTS
# ============================================================

W2 = np.array([
    [0.4],
    [0.6]
])

print("\nHidden-Output Weights")

print(W2)

# ============================================================
# OUTPUT BIAS
# ============================================================

B2 = np.array([
    0.5
])

print("\nOutput Bias")

print(B2)

# ============================================================
# OUTPUT LAYER CALCULATION
# ============================================================

output_input = (
    np.dot(
        hidden_output,
        W2
    )
    + B2
)

print("\nOutput Layer Input")

print(output_input)

# ============================================================
# SIGMOID ACTIVATION
# ============================================================

output = (
    1 /
    (
        1 +
        np.exp(-output_input)
    )
)

print("\nFinal Output")

print(output)
```

```python
# ============================================================
# MULTI LAYER PERCEPTRON
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    figsize=(14,8)
)

ax.axis("off")

layers = [
    4,
    6,
    5,
    3
]

positions = []

for layer_idx, layer_size in enumerate(layers):

    y_positions = np.linspace(
        -(layer_size-1)/2,
        (layer_size-1)/2,
        layer_size
    )

    layer_positions = []

    for y in y_positions:

        ax.scatter(
            layer_idx*2,
            y,
            s=600
        )

        layer_positions.append(
            (layer_idx*2,y)
        )

    positions.append(
        layer_positions
    )

for i in range(
    len(positions)-1
):

    for n1 in positions[i]:

        for n2 in positions[i+1]:

            ax.plot(
                [n1[0],n2[0]],
                [n1[1],n2[1]],
                linewidth=1
            )

labels = [
    "Input Layer",
    "Hidden Layer 1",
    "Hidden Layer 2",
    "Output Layer"
]

for i,label in enumerate(labels):

    ax.text(
        i*2,
        -4,
        label,
        ha="center",
        fontsize=12,
        fontweight="bold"
    )

plt.title(
    "Multi Layer Perceptron (MLP)"
)

plt.show()
```

```python
# ============================================================
# DEEP NEURAL NETWORK (DNN)
# FORWARD PROPAGATION
# ============================================================

import numpy as np

# ============================================================
# INPUT LAYER
# ============================================================

X = np.array([
    2,
    3
])

print("Input Layer")

print(X)

# ============================================================
# HIDDEN LAYER 1
# ============================================================

W1 = np.array([
    [0.5,0.2,0.4],
    [0.3,0.8,0.6]
])

B1 = np.array([
    0.1,
    0.2,
    0.1
])

Z1 = np.dot(
    X,
    W1
) + B1

A1 = np.maximum(
    0,
    Z1
)

print("\nHidden Layer 1 Output")

print(A1)

# ============================================================
# HIDDEN LAYER 2
# ============================================================

W2 = np.array([
    [0.2,0.4],
    [0.5,0.3],
    [0.7,0.6]
])

B2 = np.array([
    0.1,
    0.2
])

Z2 = np.dot(
    A1,
    W2
) + B2

A2 = np.maximum(
    0,
    Z2
)

print("\nHidden Layer 2 Output")

print(A2)

# ============================================================
# HIDDEN LAYER 3
# ============================================================

W3 = np.array([
    [0.4,0.2],
    [0.6,0.8]
])

B3 = np.array([
    0.1,
    0.1
])

Z3 = np.dot(
    A2,
    W3
) + B3

A3 = np.maximum(
    0,
    Z3
)

print("\nHidden Layer 3 Output")

print(A3)

# ============================================================
# OUTPUT LAYER
# ============================================================

W4 = np.array([
    [0.5],
    [0.7]
])

B4 = np.array([
    0.2
])

Z4 = np.dot(
    A3,
    W4
) + B4

output = (
    1 /
    (
        1 +
        np.exp(-Z4)
    )
)

print("\nFinal Output")

print(output)
```

```python
# ============================================================
# LAYER OUTPUT VISUALIZATION
# ============================================================

import pandas as pd
import plotly.express as px

layer_values = pd.DataFrame({
    "Layer":[
        "Input",
        "Hidden-1",
        "Hidden-2",
        "Hidden-3",
        "Output"
    ],
    "Value":[
        np.mean(X),
        np.mean(A1),
        np.mean(A2),
        np.mean(A3),
        output[0]
    ]
})

fig = px.bar(
    layer_values,
    x="Layer",
    y="Value",
    title="Average Activation Across Layers"
)

fig.show()
```

```python
# ============================================================
# DNN SIGNAL FLOW
# ============================================================

import plotly.graph_objects as go

labels = [
    "Input Layer",
    "Hidden Layer 1",
    "Hidden Layer 2",
    "Hidden Layer 3",
    "Output Layer"
]

fig = go.Figure(
    go.Sankey(
        node=dict(
            label=labels
        ),
        link=dict(
            source=[0,1,2,3],
            target=[1,2,3,4],
            value=[2,3,4,5]
        )
    )
)

fig.update_layout(
    title="Deep Neural Network Signal Flow"
)

fig.show()
```

```python
# ============================================================
# DNN USING TENSORFLOW
# ============================================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential()

# Input Layer

model.add(
    Dense(
        64,
        activation="relu",
        input_shape=(10,)
    )
)

# Hidden Layer 1

model.add(
    Dense(
        32,
        activation="relu"
    )
)

# Hidden Layer 2

model.add(
    Dense(
        16,
        activation="relu"
    )
)

# Hidden Layer 3

model.add(
    Dense(
        8,
        activation="relu"
    )
)

# Output Layer

model.add(
    Dense(
        1,
        activation="sigmoid"
    )
)

model.summary()
```
