```python
# ============================================================
# FEEDFORWARD NEURAL NETWORK
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    figsize=(12,6)
)

ax.axis("off")

inputs = np.linspace(
    1,
    4,
    4
)

hidden = np.linspace(
    1.5,
    3.5,
    3
)

output = [2.5]

# Input Layer

for y in inputs:

    ax.scatter(
        1,
        y,
        s=700
    )

# Hidden Layer

for y in hidden:

    ax.scatter(
        4,
        y,
        s=700
    )

# Output Layer

ax.scatter(
    7,
    output[0],
    s=900
)

# Connections

for i in inputs:

    for h in hidden:

        ax.plot(
            [1,4],
            [i,h]
        )

for h in hidden:

    ax.plot(
        [4,7],
        [h,output[0]]
    )

ax.text(1,0.3,"Input")
ax.text(4,0.3,"Hidden")
ax.text(7,0.3,"Output")

plt.title(
    "Feedforward Neural Network"
)

plt.show()
```

```python
# ============================================================
# FEEDFORWARD CALCULATION
# ============================================================

import numpy as np

X = np.array([
    2,
    3
])

W1 = np.array([
    [0.5,0.2],
    [0.3,0.8]
])

B1 = np.array([
    0.1,
    0.2
])

hidden = np.dot(
    X,
    W1
) + B1

hidden = np.maximum(
    0,
    hidden
)

print("Hidden Layer")

print(hidden)

W2 = np.array([
    [0.4],
    [0.6]
])

B2 = np.array([
    0.5
])

output = np.dot(
    hidden,
    W2
) + B2

print("Output")

print(output)
```

```python
# ============================================================
# SHALLOW NEURAL NETWORK
# ============================================================

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(
    figsize=(12,6)
)

ax.axis("off")

layers = [
    4,
    5,
    2
]

positions = []

for layer_idx, layer_size in enumerate(layers):

    y_positions = np.linspace(
        -(layer_size-1)/2,
        (layer_size-1)/2,
        layer_size
    )

    current_layer = []

    for y in y_positions:

        ax.scatter(
            layer_idx*2,
            y,
            s=600
        )

        current_layer.append(
            (layer_idx*2,y)
        )

    positions.append(
        current_layer
    )

for i in range(
    len(positions)-1
):

    for p1 in positions[i]:

        for p2 in positions[i+1]:

            ax.plot(
                [p1[0],p2[0]],
                [p1[1],p2[1]]
            )

plt.title(
    "Shallow Neural Network"
)

plt.show()
```

```python
# ============================================================
# MLP CALCULATION (FORWARD PROPAGATION)
# ============================================================

import numpy as np

# ============================================================
# INPUT LAYER
# ============================================================

X = np.array([
    2,
    3
])

print("Inputs")

print(X)

# ============================================================
# INPUT -> HIDDEN WEIGHTS
# ============================================================

W1 = np.array([
    [0.5, 0.2],
    [0.3, 0.8]
])

print("\nInput-Hidden Weights")

print(W1)

# ============================================================
# HIDDEN BIAS
# ============================================================

B1 = np.array([
    0.1,
    0.2
])

print("\nHidden Bias")

print(B1)

# ============================================================
# HIDDEN LAYER CALCULATION
# ============================================================

hidden_input = (
    np.dot(X, W1)
    + B1
)

print("\nHidden Layer Input")

print(hidden_input)

# ============================================================
# RELU ACTIVATION
# ============================================================

hidden_output = np.maximum(
    0,
    hidden_input
)

print("\nHidden Layer Output")

print(hidden_output)

# ============================================================
# HIDDEN -> OUTPUT WEIGHTS
# ============================================================

W2 = np.array([
    [0.4],
    [0.6]
])

print("\nHidden-Output Weights")

print(W2)

# ============================================================
# OUTPUT BIAS
# ============================================================

B2 = np.array([
    0.5
])

print("\nOutput Bias")

print(B2)

# ============================================================
# OUTPUT LAYER CALCULATION
# ============================================================

output_input = (
    np.dot(
        hidden_output,
        W2
    )
    + B2
)

print("\nOutput Layer Input")

print(output_input)

# ============================================================
# SIGMOID ACTIVATION
# ============================================================

output = (
    1 /
    (
        1 +
        np.exp(-output_input)
    )
)

print("\nFinal Output")

print(output)
```

```python
# ============================================================
# DEEP NEURAL NETWORK (DNN)
# FORWARD PROPAGATION
# ============================================================

import numpy as np

# ============================================================
# INPUT LAYER
# ============================================================

X = np.array([
    2,
    3
])

print("Input Layer")

print(X)

# ============================================================
# HIDDEN LAYER 1
# ============================================================

W1 = np.array([
    [0.5,0.2,0.4],
    [0.3,0.8,0.6]
])

B1 = np.array([
    0.1,
    0.2,
    0.1
])

Z1 = np.dot(
    X,
    W1
) + B1

A1 = np.maximum(
    0,
    Z1
)

print("\nHidden Layer 1 Output")

print(A1)

# ============================================================
# HIDDEN LAYER 2
# ============================================================

W2 = np.array([
    [0.2,0.4],
    [0.5,0.3],
    [0.7,0.6]
])

B2 = np.array([
    0.1,
    0.2
])

Z2 = np.dot(
    A1,
    W2
) + B2

A2 = np.maximum(
    0,
    Z2
)

print("\nHidden Layer 2 Output")

print(A2)

# ============================================================
# HIDDEN LAYER 3
# ============================================================

W3 = np.array([
    [0.4,0.2],
    [0.6,0.8]
])

B3 = np.array([
    0.1,
    0.1
])

Z3 = np.dot(
    A2,
    W3
) + B3

A3 = np.maximum(
    0,
    Z3
)

print("\nHidden Layer 3 Output")

print(A3)

# ============================================================
# OUTPUT LAYER
# ============================================================

W4 = np.array([
    [0.5],
    [0.7]
])

B4 = np.array([
    0.2
])

Z4 = np.dot(
    A3,
    W4
) + B4

output = (
    1 /
    (
        1 +
        np.exp(-Z4)
    )
)

print("\nFinal Output")

print(output)
```

# Activation Functions Cheat Sheet

| Activation Function | Mathematical Formula | Output Range | Common Usage | Advantages | Limitations |
|---------------------|----------------------|--------------|--------------|------------|-------------|
| Binary Step | f(x) = {1 if x ≥ 0, 0 otherwise} | {0,1} | Perceptron | Simple and Fast | Not Differentiable |
| Linear | f(x) = x | (-∞, +∞) | Regression Output Layer | Simple | No Non-Linearity |
| Sigmoid | f(x) = 1 / (1 + e^(-x)) | (0,1) | Binary Classification Output | Produces Probabilities | Vanishing Gradient |
| Tanh | f(x) = (e^x - e^(-x)) / (e^x + e^(-x)) | (-1,1) | Hidden Layers | Zero-Centered Output | Vanishing Gradient |
| ReLU | f(x) = max(0, x) | [0, +∞) | Hidden Layers in ANN/CNN | Computationally Efficient | Dead Neuron Problem |
| Leaky ReLU | f(x) = max(αx, x) | (-∞, +∞) | Deep Networks | Avoids Dead ReLU | Requires α Selection |
| Parametric ReLU (PReLU) | f(x) = max(ax, x) | (-∞, +∞) | Advanced CNNs | Learnable Negative Slope | Additional Parameters |
| ELU | f(x) = x if x > 0 else α(e^x - 1) | (-α, +∞) | Deep Neural Networks | Smooth Negative Values | Higher Computation Cost |
| SELU | f(x) = λ[x if x > 0 else α(e^x - 1)] | (-∞, +∞) | Self-Normalizing Networks | Stable Training | Requires Specific Initialization |
| Softplus | f(x) = ln(1 + e^x) | (0, +∞) | Smooth ReLU Alternative | Fully Differentiable | Slower than ReLU |
| Softsign | f(x) = x / (1 + |x|) | (-1,1) | Alternative to Tanh | Smooth Saturation | Less Frequently Used |
| Swish | f(x) = x · sigmoid(x) | (-∞, +∞) | EfficientNet, Modern DNNs | Better Gradient Flow | More Computational Cost |
| GELU | f(x) = x · Φ(x) | (-∞, +∞) | Transformers, GPT, BERT | State-of-the-Art Performance | Computationally Expensive |
| Softmax | f(x_i) = e^(x_i) / Σe^(x_j) | (0,1), Σ=1 | Multi-Class Classification Output | Probability Distribution | Output Layer Only |

```python
# ============================================================
# ACTIVATION FUNCTIONS VISUALIZATION
# ============================================================

import numpy as np
import plotly.graph_objects as go
from scipy.special import expit

# Input Range

x = np.linspace(
    -10,
    10,
    1000
)

# ============================================================
# ACTIVATION FUNCTIONS
# ============================================================

sigmoid = 1 / (
    1 + np.exp(-x)
)

tanh = np.tanh(x)

relu = np.maximum(
    0,
    x
)

leaky_relu = np.where(
    x > 0,
    x,
    0.01*x
)

softplus = np.log(
    1 + np.exp(x)
)

softsign = x / (
    1 + np.abs(x)
)

swish = x * sigmoid

gelu = (
    0.5 * x *
    (
        1 +
        np.tanh(
            np.sqrt(2/np.pi) *
            (
                x +
                0.044715 * x**3
            )
        )
    )
)

elu = np.where(
    x > 0,
    x,
    np.exp(x)-1
)

linear = x

# ============================================================
# PLOTLY VISUALIZATION
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=x,
        y=sigmoid,
        name="Sigmoid"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=tanh,
        name="Tanh"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=relu,
        name="ReLU"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=leaky_relu,
        name="Leaky ReLU"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=swish,
        name="Swish"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=gelu,
        name="GELU"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=elu,
        name="ELU"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=softplus,
        name="Softplus"
    )
)

fig.add_trace(
    go.Scatter(
        x=x,
        y=softsign,
        name="Softsign"
    )
)

fig.update_layout(
    title="Activation Functions Comparison",
    xaxis_title="Input",
    yaxis_title="Output",
    height=700
)

fig.show()
```

```python
# ============================================================
# LEAKY RELU
# Range : (-∞,+∞)
# Use Case : Deep Neural Networks
# ============================================================

import numpy as np
import plotly.express as px

x = np.linspace(-10,10,1000)

y = np.where(
    x>0,
    x,
    0.01*x
)

fig = px.line(
    x=x,
    y=y,
    title="Leaky ReLU"
)

fig.show()

print("Range : (-∞,+∞)")
print("Use Case : Deep Networks")
```

```python
# ============================================================
# SOFTMAX
# Range : (0,1)
# Sum of Probabilities = 1
# Use Case : Multi-Class Classification
# ============================================================

import numpy as np
import plotly.express as px

scores = np.array([
    2.5,
    1.2,
    0.8,
    3.4
])

softmax = np.exp(scores) / np.sum(
    np.exp(scores)
)

print("Softmax Probabilities")

print(softmax)

import pandas as pd

df = pd.DataFrame({
    "Class":[
        "Class A",
        "Class B",
        "Class C",
        "Class D"
    ],
    "Probability":softmax
})

fig = px.bar(
    df,
    x="Class",
    y="Probability",
    title="Softmax Probability Distribution"
)

fig.show()

print(
    "Total Probability =",
    np.sum(softmax)
)
```

```python
# ============================================================
# GELU
# Range : (-∞,+∞)
# Use Case : Transformers, LLMs
# ============================================================

import numpy as np
import plotly.express as px

x = np.linspace(-10,10,1000)

gelu = (
    0.5*x*
    (
        1+
        np.tanh(
            np.sqrt(2/np.pi)*
            (
                x+
                0.044715*x**3
            )
        )
    )
)

fig = px.line(
    x=x,
    y=gelu,
    title="GELU Activation Function"
)

fig.show()

print("Range : (-∞,+∞)")
print("Used in GPT, BERT, Transformers")
```

```python
# ============================================================
# PARAMETRIC RELU (PReLU)
# Range : (-∞,+∞)
# Use Case : Deep Neural Networks
# Alpha is Learnable
# ============================================================

import numpy as np
import plotly.express as px

alpha = 0.25

x = np.linspace(-10,10,1000)

y = np.where(
    x > 0,
    x,
    alpha*x
)

fig = px.line(
    x=x,
    y=y,
    title="Parametric ReLU (PReLU)"
)

fig.show()

print("Output Range : (-∞,+∞)")
print("Use Case : Deep Neural Networks")
```

```python
# ============================================================
# ELU
# Range : (-1,+∞)
# Use Case : Deep Networks
# Solves Dead ReLU Problem
# ============================================================

import numpy as np
import plotly.express as px

alpha = -1

x = np.linspace(-10,10,1000)

y = np.where(
    x > 0,
    x,
    alpha*(np.exp(x)-1)
)

fig = px.line(
    x=x,
    y=y,
    title="ELU Activation Function"
)

fig.show()

print("Output Range : (-1,+∞)")
print("Use Case : Deep Networks")
```

```python
# ============================================================
# SELU
# Range : (-∞,+∞)
# Use Case : Self-Normalizing Neural Networks
# ============================================================

import numpy as np
import plotly.express as px

alpha = -1.67326
scale = -1.0507

x = np.linspace(-10,10,1000)

y = scale * np.where(
    x > 0,
    x,
    alpha*(np.exp(x)-1)
)

fig = px.line(
    x=x,
    y=y,
    title="SELU Activation Function"
)

fig.show()

print("Output Range : (-∞,+∞)")
print("Use Case : Self-Normalizing Networks")
```

```python
# ============================================================
# SOFTPLUS
# Range : (0,+∞)
# Use Case : Smooth ReLU Alternative
# ============================================================

import numpy as np
import plotly.express as px

x = np.linspace(-10,10,1000)

y = np.log(
    1 + np.exp(x)
)

fig = px.line(
    x=x,
    y=y,
    title="Softplus Activation Function"
)

fig.show()

print("Output Range : (0,+∞)")
print("Use Case : Smooth ReLU")
```

```python
# ============================================================
# NORMALIZATION
# ============================================================

X_train = X_train / 255.0

X_test = X_test / 255.0

print(X_train.min())

print(X_train.max())
```

```python
# ============================================================
# ONE HOT ENCODING
# ============================================================

y_train_cat = to_categorical(
    y_train,
    10
)

y_test_cat = to_categorical(
    y_test,
    10
)

print(
    y_train_cat[0]
)
```

```python
# ============================================================
# CNN ARCHITECTURE
# ============================================================

model = Sequential()

# ------------------------------------------------------------
# BLOCK 1
# ------------------------------------------------------------

model.add(
    Conv2D(
        32,
        (3,3),
        activation="relu",
        padding="same",
        input_shape=(32,32,3)
    )
)

model.add(
    BatchNormalization()
)

model.add(
    Conv2D(
        32,
        (3,3),
        activation="relu",
        padding="same"
    )
)

model.add(
    MaxPooling2D(
        pool_size=(2,2)
    )
)

# ------------------------------------------------------------
# BLOCK 2
# ------------------------------------------------------------

model.add(
    Conv2D(
        64,
        (3,3),
        activation="relu",
        padding="same"
    )
)

model.add(
    BatchNormalization()
)

model.add(
    Conv2D(
        64,
        (3,3),
        activation="relu",
        padding="same"
    )
)

model.add(
    MaxPooling2D(
        pool_size=(2,2)
    )
)

# ------------------------------------------------------------
# BLOCK 3
# ------------------------------------------------------------

model.add(
    Conv2D(
        128,
        (3,3),
        activation="relu",
        padding="same"
    )
)

model.add(
    BatchNormalization()
)

model.add(
    MaxPooling2D(
        pool_size=(2,2)
    )
)

# ------------------------------------------------------------
# FLATTEN
# ------------------------------------------------------------

model.add(
    Flatten()
)

# ------------------------------------------------------------
# DENSE LAYERS
# ------------------------------------------------------------

model.add(
    Dense(
        256,
        activation="relu"
    )
)

model.add(
    Dropout(
        0.5
    )
)

# ------------------------------------------------------------
# OUTPUT LAYER
# ------------------------------------------------------------

model.add(
    Dense(
        10,
        activation="softmax"
    )
)

model.summary()
```

```python
# ============================================================
# COMPILE MODEL
# ============================================================

model.compile(

    optimizer="adam",

    loss="categorical_crossentropy",

    metrics=["accuracy"]

)
```

```python
# ============================================================
# TRAIN CNN
# ============================================================

history = model.fit(

    X_train,

    y_train_cat,

    validation_split=0.2,

    epochs=20,

    batch_size=64

)
```

```python
# ============================================================
# ACCURACY CURVE
# ============================================================

plt.figure(
    figsize=(10,5)
)

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.legend()

plt.title(
    "CNN Accuracy Curve"
)

plt.show()
```

```python
# ============================================================
# LOSS CURVE
# ============================================================

plt.figure(
    figsize=(10,5)
)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.legend()

plt.title(
    "CNN Loss Curve"
)

plt.show()
```

```python
# ============================================================
# EVALUATE MODEL
# ============================================================

loss, accuracy = model.evaluate(

    X_test,

    y_test_cat

)

print("Test Accuracy")

print(accuracy)
```

# **TRANSFER LEARNING MODELS**

# ============================================================
# TRANSFER LEARNING MODELS IN DEEP LEARNING
# ============================================================

Transfer Learning is a Deep Learning technique where a model
pre-trained on a large dataset (such as ImageNet) is reused
for a new task instead of training a neural network from scratch.

ImageNet Dataset:
- 14+ Million Images
- 1000 Classes

Benefits:
✓ Faster Training
✓ Less Data Requirement
✓ Better Accuracy
✓ Reduced Computational Cost

# ============================================================
# 1. VGG FAMILY
# ============================================================

VGG = Visual Geometry Group
Developed By: University of Oxford

Models:
- VGG16
- VGG19

Full Forms:
VGG16 = Visual Geometry Group 16 Layer Network
VGG19 = Visual Geometry Group 19 Layer Network

Characteristics:
- Simple Architecture
- Uses 3x3 Convolution Filters
- Easy to Understand
- High Number of Parameters

Input Size:
224 × 224 × 3

Use Cases:
✓ Image Classification
✓ Transfer Learning Education
✓ Feature Extraction
✓ Medical Imaging

Advantages:
✓ Easy to Implement
✓ Good Baseline Model

Limitations:
✗ Large Model Size
✗ Slow Training

# ============================================================
# 2. RESNET FAMILY
# ============================================================

ResNet = Residual Neural Network
Developed By: Microsoft Research

Models:
- ResNet18
- ResNet34
- ResNet50
- ResNet101
- ResNet152

Full Forms:
ResNet18  = Residual Network with 18 Layers
ResNet34  = Residual Network with 34 Layers
ResNet50  = Residual Network with 50 Layers
ResNet101 = Residual Network with 101 Layers
ResNet152 = Residual Network with 152 Layers

Key Innovation:
Residual Connections (Skip Connections)

Use Cases:
✓ Image Classification
✓ Object Detection
✓ Face Recognition
✓ Medical Imaging
✓ Industrial AI Systems

Advantages:
✓ Solves Vanishing Gradient Problem
✓ Supports Very Deep Networks
✓ High Accuracy

Most Popular:
ResNet50
